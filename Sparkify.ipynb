### create timestamp column from original timestamp column

```pyspark
df_plays.printSchema()
```

```pyspark
from pyspark.sql.functions import col, udf
from pyspark.sql.types import TimestampType
```

```pyspark
import datetime

# Define a UDF to convert a Unix timestamp to a timestamp
def from_unixtime_udf(ts):
    return datetime.datetime.fromtimestamp(ts/1000)

# Define a Spark UDF from the function
get_timestamp = udf(from_unixtime_udf, TimestampType())

# Apply the UDF to the Unix timestamp column to create a new timestamp column
df_plays_with_timestamp = df_plays.withColumn("timestamp", get_timestamp(col("ts")))

# Show the transformed DataFrame
df_plays_with_timestamp.show(10)
```

### # create datetime column from original timestamp column

```pyspark
df_plays_with_timestamp.printSchema()
```

```pyspark
from pyspark.sql.functions import from_unixtime
df_with_datetime = df_plays_with_timestamp.withColumn("datetime", from_unixtime(col("ts")/1000))


# Show the transformed DataFrame
df_with_datetime.show(10)
```

### # extract columns to create time table

```pyspark
df_with_datetime.columns
```

```pyspark
df_with_datetime.printSchema()
```

```pyspark
from pyspark.sql.functions import hour, year, month, dayofmonth, weekofyear, dayofweek

# extract columns to create time table
time_col = ['timestamp', 'datetime']
time_table = df_with_datetime[time_col]

time_table = time_table.withColumn("hour", hour(col("timestamp")))
time_table = time_table.withColumn("day", dayofmonth(col("timestamp")))
time_table = time_table.withColumn("week", weekofyear(col("timestamp")))
time_table = time_table.withColumn("month", month(col("timestamp")))
time_table = time_table.withColumn("year", year(col("timestamp")))
time_table = time_table.withColumn("weekday", dayofweek(col("timestamp")))

time_table.show(5)
```

### # write time table to parquet files partitioned by year and month

```pyspark
# write time table to parquet files partitioned by year and month
time_table.write.partitionBy("year", "month").parquet(output_data+"output-time-table")
```

### # read in song data to use for songplays table

```pyspark
song_path = output_data+"output-songs-table"
song_df = spark.read.parquet(song_path)
song_df.show(10)
```

### # extract columns from joined song and log datasets to create songplays table 

```pyspark
song_df.printSchema()
```

```pyspark
import uuid
from pyspark.sql.types import StringType

# Define a User Defined Function (UDF) to generate UUIDs
uuid_udf = udf(lambda: str(uuid.uuid4()), StringType())

# Add a new column to the DataFrame with unique UUIDs
df_with_uuid = df_with_datetime.withColumn("songplay_id", uuid_udf())

# Join with song_df to get the song_id and artist_id columns
songplays_table = df_with_uuid.join(song_df, df_with_datetime.song == song_df.title, "inner") \
                       .selectExpr("songplay_id",
                                   "timestamp as start_time",
                                   "userId as user_id",
                                   "level",
                                   "song_id",
                                   "artist_id",
                                   "sessionId as session_id",
                                   "location",
                                   "userAgent as user_agent")


songplays_table.show(20) 
```
